## 📈 Stock Market Analysis and Visualization

**Project Objective:** To analyze Apple's (AAPL) historical stock market data using Python, Pandas, NumPy, Matplotlib, and Seaborn.

This project explores stock prices, trading volume, daily returns, and market trends through data analysis and visualization.

```python
# Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
# Load Stock Market Dataset

df = pd.read_csv("dataset/AAPL_stock_data.csv")

print("Dataset loaded successfully!")

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
# Display Dataset Information

print("Dataset Information:")
df.info()

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nDataset Shape:")
print(df.shape)
```

```python
# Explore Dataset

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nStatistical Summary:")
display(df.describe())
```

```python

# Check Missing Values
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Total Missing Values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())


# Check Duplicate Records
print("\nDuplicate Records:")
print(df.duplicated().sum())


# Display Duplicate Rows
print("\nDuplicate Rows:")
display(df[df.duplicated()])


# Dataset Information
print("\nDataset Information After Checking:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

# Display Data Types of Each Column
print("Data Types of Each Column:")
print(df.dtypes)

# Check Dataset Information
print("\nComplete Dataset Information:")
df.info()

# Count Data Types
print("\nData Type Counts:")
print(df.dtypes.value_counts())
```

```python
# Check Unnamed: 0 Column

print(df["Unnamed: 0"].head(10))

print("\nSample Date Values:")
print(df["Unnamed: 0"].tail(5))
```

```python

# Rename Unnamed: 0 Column to Date
df.rename(columns={"Unnamed: 0": "Date"}, inplace=True)

# Convert Date Column into Datetime Format
df["Date"] = pd.to_datetime(df["Date"])

# Remove Duplicate Records
df.drop_duplicates(inplace=True)

# Reset Index
df.reset_index(drop=True, inplace=True)

# Display Cleaned Dataset
print("Data Cleaning Completed Successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nUpdated Column Names:")
print(df.columns.tolist())

print("\nUpdated Data Types:")
print(df.dtypes)

print("\nCleaned Dataset:")
display(df.head())
```

```python

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nDataset Information:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

df = df.sort_values(by="Date", ascending=True)

# Reset Index
df.reset_index(drop=True, inplace=True)

print("Dataset Sorted Successfully!")

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())

print("\nDate Range:")
print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())
```

```python

print("Statistical Analysis of Stock Data:")

display(df[["open", "high", "low", "close", "adjclose", "volume"]].describe())

print("\nAverage Stock Prices:")
print("Average Open Price:", df["open"].mean())
print("Average High Price:", df["high"].mean())
print("Average Low Price:", df["low"].mean())
print("Average Close Price:", df["close"].mean())
print("Average Adjusted Close Price:", df["adjclose"].mean())

print("\nAverage Trading Volume:")
print(df["volume"].mean())
```

```python
# Stock Closing Price Visualization

plt.figure(figsize=(14, 6))

plt.plot(
    df["Date"],
    df["close"],
    color="blue",
    linewidth=1.8,
    label="Closing Price"
)

plt.fill_between(
    df["Date"],
    df["close"],
    color="skyblue",
    alpha=0.25
)

plt.title("Apple (AAPL) Historical Closing Stock Price",
          fontsize=16, color="darkblue")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Stock Price (USD)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Trading Volume Visualization

plt.figure(figsize=(14, 6))

plt.plot(
    df["Date"],
    df["volume"],
    color="green",
    linewidth=1,
    label="Trading Volume"
)

plt.fill_between(
    df["Date"],
    df["volume"],
    color="lightgreen",
    alpha=0.3
)

plt.title("Apple (AAPL) Historical Trading Volume",
          fontsize=16, color="darkgreen")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Trading Volume (Shares)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Moving Average Calculation

# Calculate 20 Days Moving Average
df["MA20"] = df["close"].rolling(window=20).mean()

# Calculate 50 Days Moving Average
df["MA50"] = df["close"].rolling(window=50).mean()

print("Moving Averages Calculated Successfully!")

print("\nUpdated Dataset:")
display(df[["Date", "close", "MA20", "MA50"]].tail(10))
```

```python
# Moving Average Visualization

plt.figure(figsize=(14, 7))

plt.plot(
    df["Date"],
    df["close"],
    color="blue",
    linewidth=1.2,
    label="Closing Price"
)

plt.plot(
    df["Date"],
    df["MA20"],
    color="orange",
    linewidth=1.5,
    label="20 Days Moving Average"
)

plt.plot(
    df["Date"],
    df["MA50"],
    color="red",
    linewidth=1.5,
    label="50 Days Moving Average"
)

plt.title("Apple (AAPL) Stock Price with Moving Averages",
          fontsize=16, color="darkblue")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Stock Price (USD)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Daily Returns Calculation

df["Daily_Return"] = df["adjclose"].pct_change() * 100

print("Daily Returns Calculated Successfully!")

print("\nDaily Returns:")
display(df[["Date", "adjclose", "Daily_Return"]].tail(10))

print("\nDaily Returns Statistics:")
display(df["Daily_Return"].describe())
```

```python
# Daily Returns Visualization

plt.figure(figsize=(14, 6))

plt.plot(
    df["Date"],
    df["Daily_Return"],
    color="blue",
    linewidth=0.8,
    label="Daily Returns"
)

# Zero Reference Line
plt.axhline(
    y=0,
    color="black",
    linestyle="--",
    linewidth=1
)

# Positive Returns
plt.fill_between(
    df["Date"],
    df["Daily_Return"],
    0,
    where=(df["Daily_Return"] >= 0),
    color="green",
    alpha=0.3,
    label="Positive Returns"
)

# Negative Returns
plt.fill_between(
    df["Date"],
    df["Daily_Return"],
    0,
    where=(df["Daily_Return"] < 0),
    color="red",
    alpha=0.3,
    label="Negative Returns"
)

plt.title("Apple (AAPL) Daily Stock Returns",
          fontsize=16, color="darkblue")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Daily Return (%)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
#  Correlation Analysis

correlation = df[
    ["open", "high", "low", "close",
     "adjclose", "volume", "MA20", "MA50", "Daily_Return"]
].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Apple (AAPL) Stock Data Correlation Heatmap",
          fontsize=16, color="darkblue")

plt.tight_layout()
plt.show()
```

```python
# Daily Returns Distribution

plt.figure(figsize=(12, 6))

sns.histplot(
    df["Daily_Return"].dropna(),
    bins=100,
    kde=True,
    color="purple",
    edgecolor="black",
    alpha=0.6
)

plt.title("Distribution of Apple (AAPL) Daily Returns",
          fontsize=16, color="darkblue")

plt.xlabel("Daily Returns (%)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.axvline(
    df["Daily_Return"].mean(),
    color="red",
    linestyle="--",
    linewidth=2,
    label="Average Daily Return"
)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Open vs Close Price Analysis

plt.figure(figsize=(10, 6))

plt.scatter(
    df["open"],
    df["close"],
    color="purple",
    alpha=0.4,
    s=15,
    label="Stock Prices"
)

plt.title("Apple (AAPL) Opening vs Closing Price",
          fontsize=16, color="darkblue")

plt.xlabel("Opening Price (USD)", fontsize=12)
plt.ylabel("Closing Price (USD)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Best and Worst Trading Days

# Top 5 Best Trading Days
best_days = df.nlargest(5, "Daily_Return")

# Top 5 Worst Trading Days
worst_days = df.nsmallest(5, "Daily_Return")

print("Top 5 Best Trading Days:")

display(
    best_days[["Date", "close", "Daily_Return"]]
)

print("\nTop 5 Worst Trading Days:")

display(
    worst_days[["Date", "close", "Daily_Return"]]
)
```

```python
# Cumulative Returns Analysis

# Calculate Cumulative Returns
df["Cumulative_Return"] = (
    (1 + df["Daily_Return"] / 100).cumprod() - 1
) * 100

print("Cumulative Returns Calculated Successfully!")

print("\nLatest Cumulative Return:")
print(
    round(df["Cumulative_Return"].iloc[-1], 2),
    "%"
)

print("\nUpdated Dataset:")
display(
    df[["Date", "adjclose", "Daily_Return",
        "Cumulative_Return"]].tail(10)
)
```

```python
# Cumulative Returns Visualization

plt.figure(figsize=(14, 6))

plt.plot(
    df["Date"],
    df["Cumulative_Return"],
    color="purple",
    linewidth=1.8,
    label="Cumulative Returns"
)

plt.fill_between(
    df["Date"],
    df["Cumulative_Return"],
    color="violet",
    alpha=0.25
)

plt.title("Apple (AAPL) Cumulative Returns Analysis",
          fontsize=16, color="darkblue")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Cumulative Returns (%)", fontsize=12)

plt.legend()
plt.grid(color="gray", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Maximum Drawdown Analysis

# Calculate Running Peak
df["Peak"] = df["adjclose"].cummax()

# Calculate Drawdown
df["Drawdown"] = (
    (df["adjclose"] / df["Peak"]) - 1
) * 100

# Find Maximum Drawdown
max_drawdown = df["Drawdown"].min()

# Find Maximum Drawdown Date
max_drawdown_date = df.loc[
    df["Drawdown"].idxmin(), "Date"
]

print("Maximum Drawdown Analysis")

print("\nMaximum Drawdown:")
print(round(max_drawdown, 2), "%")

print("\nMaximum Drawdown Date:")
print(max_drawdown_date)

print("\nLatest Drawdown:")
print(round(df["Drawdown"].iloc[-1], 2), "%")

print("\nDrawdown Records:")
display(
    df[["Date", "adjclose", "Peak", "Drawdown"]].tail(10)
)
```

## Maximum Drawdown Visualization

This visualization shows the decline in Apple's stock price from its previous highest value over time.

```python
# Maximum Drawdown Visualization

plt.figure(figsize=(14, 6))

plt.plot(df["Date"], df["Drawdown"],
         color="red", linewidth=1.5, label="Drawdown")

plt.fill_between(df["Date"], df["Drawdown"], 0,
                 color="red", alpha=0.2)

plt.title("Apple (AAPL) Maximum Drawdown Analysis",
          fontsize=16, color="darkred")

plt.xlabel("Year", fontsize=12)
plt.ylabel("Drawdown (%)", fontsize=12)

plt.axhline(y=0, color="black", linestyle="--")

plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## Yearly Returns and Volatility Analysis

This section analyzes Apple's yearly stock returns and annualized volatility to understand its long-term performance and risk.

```python
# Yearly Returns and Volatility Analysis

# Sort Dataset by Date
df = df.sort_values("Date").reset_index(drop=True)

# Calculate Yearly Closing Price
yearly_close = df.groupby(df["Date"].dt.year)["adjclose"].last()

# Calculate Yearly Returns
yearly_returns = yearly_close.pct_change() * 100

# Calculate Annualized Volatility
yearly_volatility = (
    df.groupby(df["Date"].dt.year)["Daily_Return"].std() * np.sqrt(252)
)

# Combine Results
yearly_analysis = pd.DataFrame({
    "Yearly_Return (%)": yearly_returns,
    "Annualized_Volatility (%)": yearly_volatility
})

# Display Last 10 Years
print("Yearly Returns and Volatility Analysis:")
display(yearly_analysis.tail(10))

# Average of Complete Dataset
print("\nAverage Yearly Return (Complete Dataset):")
print(yearly_returns.mean())

print("\nAverage Annualized Volatility (Complete Dataset):")
print(yearly_volatility.mean())

# Average of Last 10 Years
print("\nAverage Yearly Return (Last 10 Years):")
print(yearly_returns.tail(10).mean())

print("\nAverage Annualized Volatility (Last 10 Years):")
print(yearly_volatility.tail(10).mean())
```

## Monthly Returns Analysis

This section analyzes Apple's monthly stock returns to identify seasonal trends and patterns in stock market performance.

```python
# Monthly Returns Analysis

# Calculate Monthly Closing Prices
monthly_close = df.groupby(
    df["Date"].dt.to_period("M")
)["adjclose"].last()

# Calculate Monthly Returns
monthly_returns = monthly_close.pct_change() * 100

# Create Monthly Returns DataFrame
monthly_analysis = monthly_returns.reset_index()

monthly_analysis.columns = ["Month", "Monthly_Return (%)"]

# Extract Year
monthly_analysis["Year"] = monthly_analysis["Month"].dt.year

# Extract Month Name (Corrected)
monthly_analysis["Month_Name"] = pd.to_datetime(
    monthly_analysis["Month"].astype(str)
).dt.month_name()

# Create Monthly Returns Pivot Table
monthly_pivot = monthly_analysis.pivot(
    index="Year",
    columns="Month_Name",
    values="Monthly_Return (%)"
)

# Arrange Months in Correct Order
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_pivot = monthly_pivot.reindex(columns=month_order)

print("Monthly Returns Analysis:")
display(monthly_pivot.tail(10))
```

## Monthly Returns Heatmap Visualization

This heatmap represents Apple's monthly stock returns across different years. It helps identify positive and negative return patterns using color variations.

```python
# Monthly Returns Heatmap

plt.figure(figsize=(16, 8))

sns.heatmap(
    monthly_pivot,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    center=0,
    linewidths=0.5,
    linecolor="white",
    cbar_kws={"label": "Monthly Returns (%)"}
)

plt.title(
    "Apple (AAPL) Monthly Returns Heatmap",
    fontsize=18,
    color="darkblue",
    fontweight="bold"
)

plt.xlabel("Month", fontsize=12)
plt.ylabel("Year", fontsize=12)

plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

## Average Monthly Returns Analysis

This section analyzes the average stock returns for each month to identify seasonal trends in Apple's stock performance.

```python
# Average Monthly Returns Analysis

# Calculate Average Return for Each Month
average_monthly_returns = monthly_analysis.groupby(
    "Month_Name"
)["Monthly_Return (%)"].mean()

# Arrange Months in Correct Order
average_monthly_returns = average_monthly_returns.reindex(month_order)

print("Average Monthly Returns:")
display(average_monthly_returns)

# Visualize Average Monthly Returns
plt.figure(figsize=(14, 6))

colors = [
    "green" if value >= 0 else "red"
    for value in average_monthly_returns
]

plt.bar(
    average_monthly_returns.index,
    average_monthly_returns.values,
    color=colors,
    edgecolor="black"
)

plt.axhline(0, color="black", linewidth=1)

plt.title(
    "Apple (AAPL) Average Monthly Returns",
    fontsize=16,
    color="darkblue",
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("Average Return (%)")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

## Yearly Returns Visualization

This section visualizes Apple's yearly stock returns to compare annual performance and identify the best and worst performing years.

```python
# Yearly Returns Visualization

plt.figure(figsize=(14, 6))

colors = [
    "green" if value >= 0 else "red"
    for value in yearly_analysis["Yearly_Return (%)"]
]

plt.bar(
    yearly_analysis.index,
    yearly_analysis["Yearly_Return (%)"],
    color=colors,
    edgecolor="black"
)

plt.axhline(0, color="black", linewidth=1)

plt.title(
    "Apple (AAPL) Yearly Stock Returns",
    fontsize=16,
    color="darkblue",
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Yearly Returns (%)")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
